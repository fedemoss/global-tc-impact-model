# Case studies

Four events, each as reported impact beside the two models' predictions over
the same administrative units, with the storm's best track.

The extent of a case is chosen rather than derived: a storm is legible only
when the frame shows where it came from, which is not the same as the area it
damaged. A case left out of `EXTENTS` is framed on the units it affected.


```python
import os
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import matplotlib.pyplot as plt
import pandas as pd

from visualization.lib import RUN_SUFFIX

# Figures are written here as well as shown, so each run keeps its own set.
FIGDIR = Path.cwd() / f"figs{RUN_SUFFIX}"
FIGDIR.mkdir(exist_ok=True)
print(f"run suffix {RUN_SUFFIX!r}; figures -> {FIGDIR}")
```

```python
from visualization.lib import plot_case_studies

# (DisNo., IBTrACS sid, caption)
CASES = [
    ("2016-0355-LCA", "2016273N13300", "Saint Lucia, Tropical Storm Matthew (2016)"),
    ("2005-0585-MEX", "2005289N18282", "Mexico, Hurricane Wilma (2005)"),
    ("2019-0165-MOZ", "2019112S10053", "Mozambique, Cyclone Kenneth (2019)"),
    ("2013-0437-PHL", "2013301N13142", "Philippines, Typhoon Krosa/Vinta (2013)"),
]

EXTENTS = {
    # Close on Saint Lucia, keeping its neighbours for orientation.
    "2016-0355-LCA": (-61.5, 13.1, -59.3, 15.0),
    # Wide enough to show Wilma arriving from the Caribbean and leaving to the
    # north-east, which is what makes the track legible.
    "2005-0585-MEX": (-95.0, 14.0, -75.0, 31.0),
}
```

```python
fig = plot_case_studies(CASES, suffix=RUN_SUFFIX, extents=EXTENTS,
                        savefig=FIGDIR / "case_study_maps.png")
plt.show()
```
