# SHAP, by stage

Each stage is explained over the rows it actually scores. Stage 1 ranks every
grid cell; stage 2 only ever sees the cells stage 1 flagged, so explaining the
regressor over all cells would describe inputs it never meets.


```python
import os
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import matplotlib.pyplot as plt
import pandas as pd

from visualization.lib import RUN_SUFFIX

# Figures are written here as well as shown, so each run keeps its own set.
FIGDIR = Path.cwd() / f"figs{RUN_SUFFIX}"
FIGDIR.mkdir(exist_ok=True)
print(f"run suffix {RUN_SUFFIX!r}; figures -> {FIGDIR}")
```

```python
from src.config import OUTPUT_DIR
from visualization.lib import plot_shap_summary, shap_inputs

RUN = f"2-stage-XGBoost_global_grid{RUN_SUFFIX}"
run_folder = OUTPUT_DIR / "model_output" / RUN

(clf, X_clf), (reg, X_reg), threshold = shap_inputs(run_folder)
print(f"stage-1 cut {threshold}; {len(X_clf):,} cells sampled, "
      f"{len(X_reg):,} flagged ({100 * len(X_reg) / len(X_clf):.1f}%)")
```

```python
fig, values = plot_shap_summary(clf, X_clf,
                               savefig=FIGDIR / "shap_summary_classifier.png")
plt.show()
```

```python
fig, values = plot_shap_summary(reg, X_reg,
                               savefig=FIGDIR / "shap_summary_regressor.png")
plt.show()
```
