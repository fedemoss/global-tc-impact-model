# Where to put the high-impact boundary

The 15% cut that separates high impact from low is a choice, and the figure
shows what it costs. The upper panel is how well the model identifies the
high-impact class as the cut moves; the lower panel is how much of the data
is left above it.

The two are read together: a score that improves while the class it describes
shrinks to a handful of units is not a better model.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import matplotlib.pyplot as plt

from visualization.lib import RUN_SUFFIX

FIGDIR = Path.cwd() / f"figs{RUN_SUFFIX}"
FIGDIR.mkdir(exist_ok=True)
print(f"run suffix {RUN_SUFFIX!r}; figures -> {FIGDIR}")
```

```python
from visualization.lib import (HIGH_IMPACT_SWEEP, load_variant,
                               plot_threshold_sweep, threshold_sweep)

model = load_variant("2-stage-XGBoost_global_adm1").dropna(
    subset=["reported", "predicted"])
print(f"{len(model):,} ADM1-event observations over "
      f"{model['DisNo.'].nunique()} events")

sweep = threshold_sweep(model, HIGH_IMPACT_SWEEP)
sweep.round(3)
```

```python
fig = plot_threshold_sweep(
    sweep, savefig=FIGDIR / "binary_metrics_threshold_high_impact.png")
plt.show()
```
